```python
paragraph = """Hi good Afternoon,

Myself  Jithendra Varma, I have 6 years of experience and  I have been working as a GCP Data Engineer and also I have done GCP PDE Certification, GCP ACE certification, GCP PCA Certification and also as a Data Engineer, I have been involved in several high-impact projects, most notably for global clients in retail domain. and My primary responsibility is to  design, development, and optimization of data pipelines in Google Cloud Platform (GCP)"""
```

```python
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
import nltk
nltk.download('stopwords')
```

```python
stemmer = PorterStemmer()
sentenses = nltk.sent_tokenize(paragraph)
```

```python
#apply stopwords and filter and stemming
for i in range(len(sentenses)):
    words = nltk.word_tokenize(sentenses[i])
    words =[stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentenses[i] = ' '.join(words) # converting all the list words into sentence after stemming and filtering stopwords
```

```python
sentenses
```

```python
from nltk.stem import SnowballStemmer
snowball_stemmer = SnowballStemmer("english")

for i in range(len(sentenses)):
    words = nltk.word_tokenize(sentenses[i])
    words =[snowball_stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentenses[i] = ' '.join(words) # converting all the list words into sentence after stemming and filtering stopwords
```

```python
sentenses
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
import nltk

for i in range(len(sentenses)):
    words = nltk.word_tokenize(sentenses[i])
    nltk.download('wordnet')
    sentenses[i] = ' '.join(words) # converting all the list words into sentence after stemming and filtering stopwords
```

```python
for i in range(len(sentenses)):
    words = nltk.word_tokenize(sentenses[i])
    words =[lemmatizer.lemmatize(word.lower()) for word in words if word not in set(stopwords.words('english'))]
    sentenses[i] = ' '.join(words) # converting all the list words into sentence after lemmatization and filtering stopwords
```

```python
from nltk.corpus import stopwords
nltk.download('averaged_perceptron_tagger')
sentenses = nltk.sent_tokenize(paragraph)
```

```python
for i in range(len(sentenses)):
    words = nltk.word_tokenize(sentenses[i])
    words =[word for word in words if word not in set(stopwords.words('english'))]
    #sentenses[i] = ' '.join(words) # converting all the list words into sentence after stemming and filtering stopwords
    nltk.download('averaged_perceptron_tagger_eng')
    pos_tag = nltk.pos_tag(words)
    print(pos_tag)
```
